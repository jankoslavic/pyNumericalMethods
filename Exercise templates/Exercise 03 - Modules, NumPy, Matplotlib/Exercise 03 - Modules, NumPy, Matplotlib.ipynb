#### A simple example:

*Important:* `matplotlib` plots points - pairs `(x, y)`! 

We need to prepare numerical *arrays of points* on the abscissa and the ordinate.

```python
x = np.linspace(0, 10, 100)
```

```python
plt.plot(x, np.sin(x**1.2), 'k.')
```

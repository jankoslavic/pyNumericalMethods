## Basics of object-oriented programming

```python
# Example 1: the RandomPoints class
class RandomPoints:
    """A class that plots random points."""

    def __init__(self, N=5, x_range=(0,1), y_range=(0,1), style='ro'):
        """Class constructor."""
        self.N = N
        self.style = style
        self.xrange = x_range
        self.yrange = y_range

    def draw(self):
        """Class method that computes and plots random points."""
        self.X = np.linspace(self.xrange[0], self.xrange[1], self.N)
        self.Y = np.random.rand(self.N)*(self.yrange[1]-self.yrange[0]) + self.yrange[0]

        plt.plot(self.X, self.Y, self.style)
```

An example of using different options when constructing an object of the ``RandomPoints`` class:

```python
object_2 = RandomPoints(50, (10, 100), (0,5), style='c*')
object_2.draw()
```

An example of a class that extends the functionality of the RandomPoints class so that we do not need to call the draw method, and that also draws the grid:

```python
# Example 2: class inheritance
class RandomPoints2(RandomPoints):
    """A class derived from the ``RandomPoints`` class that plots the points automatically."""

    def __init__(self, **kwargs):
        super().__init__(**kwargs)
        self.draw()
        plt.grid()
```

```python
# Example 3: converting a symbolic expression into a numerical function with sym.lambdify:
import sympy as sym
import numpy as np
import matplotlib.pyplot as plt

symbol_x = sym.symbols('x')
symbolic_expression = sym.sin(symbol_x**2)
sym.plot(symbolic_expression, (symbol_x, 0, 3), title='symbolic')
numpy_function = sym.lambdify(symbol_x, symbolic_expression, 'numpy')
x_num_array = np.linspace(0, 3, 100)
plt.plot(x_num_array, numpy_function(x_num_array), 'r.')
plt.title('numerical')
plt.show() # needed when %matplotlib inline is not used
```

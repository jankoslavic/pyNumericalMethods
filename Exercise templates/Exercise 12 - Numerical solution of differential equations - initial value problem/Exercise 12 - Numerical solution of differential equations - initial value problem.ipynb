### Euler's method

````python
# Euler function
def euler(fun, x, y0, *args, **kwargs):
    """
    Euler's method for solving ordinary differential equations.
    
    :param fun: function of the first derivative y'(x, y)
    :param x: independent variable
    :param y0: initial values of the dependent variable y(x=0)
    :return: y: values of the dependent variable at the given values of x
    """
    y = np.zeros((len(x), len(y0)))
    y[0] = np.array(y0).copy()
    h = x[1] - x[0]
    for i, xi in enumerate(x[:-1]):
        y[i+1] = y[i] + fun(xi, y[i], *args, **kwargs) * h
    return y
````

#### Advanced use of `scipy.integrate.solve_ivp` - controlling the integration:

````python
# Example of using `scipy.integrate.solve_ivp` with an event:
def upward_cannon(t, y):
    return [y[1], -0.5]
def hit_ground(t, y):
    return y[1]
hit_ground.terminal = True
hit_ground.direction = -1
sol = solve_ivp(upward_cannon, [0, 100], [0, 10], events=hit_ground)
print(sol.t_events)
````

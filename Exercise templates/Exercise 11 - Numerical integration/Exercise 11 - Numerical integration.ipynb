### Gaussian integration method (Gaussian quadrature)

##### Using ``scipy.special.legendre``

```python
# Using `scipy.special.legendre` for Gaussian integration:
from scipy.special import legendre
n_nodes = 1
xi = legendre(n_nodes).weights[:,0]
Ai = legendre(n_nodes).weights[:,1]
```

```python
# Transformation of integration limits for Gaussian integration:
def gauss_limits(xi, a, b):
    """
    Transform the Gaussian integration limits from `xi` to the interval [a, b].
    """
    return (b+a)/2 + (b-a)/2*xi
```

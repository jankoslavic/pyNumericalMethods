## The print function, string formatting

### Formatting numbers

```python
# Printing numbers
num = 189.19341
formats = ['', 'g', 'f', '+f', '.2f', '3.2f', '8.2f', '<8.2f', '^8.2f', '9.0f', '.2e', '8.2e', '.5e']
print(f'Let us print the number {num}:')
print('format\t|output|')
print(20*'-')
for f in formats:
    print(f'{f!r:}\t|{num:{f:s}}|')
```

## Functions

```python
# First function
def first_function(a, b, c=1):
    """
    A function that takes the values `a`, `b` and `c` and prints them.
    Returns the value `a*b*c`.
    """
    print(f'a = {a}')
    print(f'b = {b}')
    print(f'c = {c}')
    return a*b*c
```

## File handling

```python
# File handling
path = 'file.txt'
my_file = open(path, mode='w+')
print('Current file contents:')
print(my_file.readlines(), '\n')

my_file.write('Hello, file!\n')
print('Line 2', file=my_file)
print('Writing to the file...')

print('New file contents:', end='\t')
my_file.seek(0)
print(my_file.readlines())
my_file.close()
```

### Stripping and replacing characters in a string, splitting strings

```python
# Working with strings (split, replace...)
list_of_lines = [f'{k:s}\t{v:}\n' for k, v in data_dict.items()]
s1 = [line.rstrip() for line in list_of_lines]
s2 = [line.replace('\t', ' = ') for line in s1]
s3 = [line.split(' = ') for line in s2]
```

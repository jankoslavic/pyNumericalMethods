### Polynomial interpolation - Lagrange method

```python
def lagrange_i(x, x_int, i):
    """ Returns the values of the i-th Lagrange polynomial

   x: list of x values of the points being interpolated
   x_int: independent variable (points at which the values are sought)
   i: index of the Lagrange polynomial
   """
    Li = 1.0
    for j in range(len(x)):

        if j != i:
            Li *= (x_int-x[j]) / (x[i]-x[j])
    return Li
```

```python
x = np.linspace(0,1,4)
y = np.random.rand(len(x))
x_int = np.linspace(np.min(x), np.max(x), 50)
plt.figure()
plt.axhline(y=0, color='k')
plt.axhline(y=1, color='k')
plt.plot(x, y, 'ro')
for i in range(len(x)):
    plt.axvline(x=x[i], color='k')
    plt.plot(x_int, lagrange_i(x, x_int, i), label=r'$l_{!s}$'.format(i))
plt.title('Lagrange polynomials')
plt.legend()
```

### Uniqueness of the solution

```python
def has_solution(A, b):
    """ Based on A and b, prints whether the system of equations Ax=b is solvable. """
    D = np.linalg.det(A)
    r_A = np.linalg.matrix_rank(A)
    r_Ab = np.linalg.matrix_rank(np.column_stack((A, b)))
    print(f'\tdet: {D}\n\trank(A): {r_A}\n\trank(Ab): {r_Ab}')
    if not D:
        if r_A != r_Ab:
            print('No solution.')
        else:
            print('Infinitely many solutions.')
    else:
        print('The system is solvable.')
```

**Question 8:** Prepare a function ``gauss_preparation`` that takes the matrix $\mathbf{A}$ and the vector $\mathbf{b}$ as arguments, builds the augmented coefficient matrix $\mathbf{[A|b]}$ and, for each pivot row ``i``, $0 \leq i \leq n-1$, computes and prints the constant $\lambda$ for all rows below the pivot row ($i+1 \leq j \leq n$) ($n$ is the number of rows of $\mathbf{A}$).
$$ \lambda = \frac{a_{ji}}{a_{ii}}$$

**Question 9:** Prepare a new function ``gauss_elimination`` that, if the coefficient $a_{ii}$ is nonzero, performs the replacement:

$$E_j \leftarrow E_j - \lambda E_i$$

and returns the result of the elimination. Do not write all the code from scratch; reuse the code of the existing function ``gauss_preparation``!

```python
def gauss_elimination(A, b):
    """ Uses Gaussian elimination to obtain the upper triangular matrix for solving the SLE. """
    Ab = np.column_stack((A, b))
    for i, pivot in enumerate(Ab[:-1]):
        for row in Ab[i+1:]:
            m = row[i] / pivot[i]
            row -= m*pivot
    return Ab
```

**Question 10:** Prepare a function ``gauss_solution`` that, based on the result of the ``gauss_elimination`` function, computes the elements of the vector of unknowns $\mathbf{x}$ using back substitution. Hint: you can reverse the order of the rows of a matrix with ``A[::-1]``.

**Back substitution:**
    A system with an upper triangular augmented coefficient matrix is solved "from the bottom up":

   $$x_i = \frac{b_i - \sum_{j=i+1}^{n}a_{ij} x_j}{a_{ii}}$$

```python
def gauss_solution(Ab):
    """
	Back substitution for solving the eliminated system.
	Food for thought: the code differs from the lecture, but the result is the same!
	"""
    n = Ab.shape[0]
    x = np.zeros(n)
    A = Ab[:, :-1]
    b = Ab[:, -1]
    for i in range(n-1, -1, -1):
        x[i] = (b[i] - np.dot(A[i, i+1:], x[i+1:])) / A[i, i]
    return x
```

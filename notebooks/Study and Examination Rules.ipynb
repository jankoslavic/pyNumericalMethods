* *solving differential equations*.

### What we grade in the individual project

The purpose of the project is to demonstrate the **appropriate selection and use of numerical methods, to comment on how they work, and to evaluate the obtained results**. We grade four areas; the numerical methods are the most important.

<details>
<summary><b>Grading criteria by area</b></summary>

**1. Numerical methods** (each of the six prescribed sections is graded separately)

* *Correctness and appropriateness of the method:*
    * the problem is solved **numerically**; a purely symbolic solution (`sympy`) or just writing down the equations is not enough,
    * the method suits the problem (e.g., solve an overdetermined system with the least-squares method, do not approximate a periodic signal with a polynomial, convert a higher-order differential equation into a system of first-order equations, etc.),
    * where possible, the result is **verified**: by comparison with an analytical solution, another numerical method, or an existing function from `numpy`/`scipy`,
    * using existing functions (`numpy`, `scipy`) is encouraged; your own implementation must be justified and compared with an existing function.
* *Reasoning and comments*: for every method you use, explain in writing:
    * **why** you chose this method and not another (e.g., Newton's method or bisection? why LU decomposition? why splines and not a polynomial? etc.),
    * **briefly**: what the assumptions of the method are and what its **limitations** are (accuracy, convergence, order of the method, choice of step size, end points in differentiation ...),
    * what the result means for your physical problem (**interpretation**).
    * **Important:** in the project, **do not repeat derivations or the theoretical background of the methods**, etc., unless this is essential for understanding or running your code. Theory is the subject of the written part of the exam.

**2. Report and code**

* *Report:*
    * basic information (first and last name, enrollment number, date) and an authorship statement,
    * problem definition with a description and a **figure** of the problem, and a list of sources,
    * sections following the [**template**](https://github.com/jankoslavic/pyNumericalMethods/tree/main/project/template) (see also the [**example project**](https://github.com/jankoslavic/pyNumericalMethods/tree/main/project/example)); the headings make it clear where each numerical topic is covered,
    * explanations in `markdown` cells, equations written in LaTeX,
    * results presented graphically (`matplotlib`), with labeled axes and units.
* *Code:*
    * after `Restart` $\to$ `Run All` the report runs **without errors**, file paths are relative (all required files are included, the code works anywhere, not only on your computer),
    * the code is readable and in a sensible order,
    * functions used more than once are collected in **your own module** (`.py`), which you import into the report,
    * functions have **docstrings** (description, parameters, return value) and meaningful comments,
    * additional packages are listed with installation instructions (e.g., `requirements.txt`).

**3. Personal approach and creative addition**

* *Personal approach:* the problem is a meaningful whole and not a set of unrelated tasks; the numerical content arises from the problem; the analysis is thorough (e.g., influence of parameters, comparison of methods, error estimate); the results are critically evaluated (are they physically meaningful?).
* *Creative addition:* content that was not covered in lectures and exercises (see the list below). We grade the quality of the implementation, not just its presence.

**4. Code tests and/or user interface**

At least one test of your own functions (e.g., with `numpy.testing` or `pytest`) or a simple user interface. The test must be run in the report and must pass.

**The most common shortcomings** of past projects that led to lower grades:
* a method is used, but it is not explained why this one, how it works, and how accurate the result is,
* a section is solved only symbolically,
* an own implementation of a method without justification and without comparison with an existing function,
* results without interpretation,
* the code does not run because of wrong paths or missing packages.

</details>

<details>
<summary><b>Plagiarism in the individual project</b></summary>

All submitted individual projects are checked with a plagiarism-detection system.

The system checks the similarity of submitted documents against all other projects from the current and past academic years. Any matching of text and of procedures in the submitted code is flagged as plagiarism (even if it is used in a different context, a different order, with renamed variables, etc.). The content of the lecture templates published in this repository is excluded from the check.

If a significant part of the content of your project matches another project, we cannot grade it as your own independent work. This means that you **have not completed the exercises and do not meet the requirements to take the exam**.

</details>


### Project grading procedure

<details>
<summary><b>The project is graded in two steps: automated review and oral defense</b></summary>

1. **Automated review of the submitted reports.** 

    All submitted reports are first reviewed automatically: we check whether the report runs without errors and whether it contains all the prescribed content, and we compare it with the other projects (see *Plagiarism* above).
2. **Oral defense** with the course instructor, about **10 minutes per student**. 

    The student briefly presents the project (the problem, the chosen methods, the key results), and the instructor then asks a few questions about the details, e.g.: why you chose a particular method, how your code works, how you verified the result.

The final project grade is based on both steps. A project that the student cannot explain at the defense cannot be graded as independent work. 

The defense of individual projects will take place approximately 1 week after submission.

</details>



### Homework (moj.ladisk.si)

We assign weekly homework in the [moj.ladisk.si](https://moj.ladisk.si/) system. You solve the homework in the Jupyter Notebook environment.

<details>
<summary><b>General instructions for solving</b></summary>

* Save the files you download from the web as a `.zip` archive (if your browser does not add the `.zip` extension to the file, do it yourself).

* **Fully extract** the archive of files you download from the web.
* In the folder with the extracted files, open a command window and run the command `jupyter notebook`.
* Run the cells in the `.ipynb` homework file in order (including the cells at the very top of the file!).
* When solving, results are sent to the moj.ladisk.si online system and are automatically checked on the fly. An **internet connection is required** when solving.


We have also prepared [video instructions for solving](https://youtu.be/wedX7AdORvY).

</details>



<details>
<summary><b>Time limit for solving homework</b></summary>

Homework has a defined *solving window* time:
* Until the solving window opens, **early solving of the assignment without a time limit** is possible.

* During the solving window, the allowed time for solving the assignment is limited. The time limit starts from the moment you first open the assignment *within the solving window*.
* After the window closes, the assignment closes and solving **is no longer possible**.

We advise you to *solve as much of the assignment as possible during the early-solving period, before the solving window opens*.

</details>



<details>
<summary><b>Using a Joker to postpone an obligation</b></summary>

Each student has two Jokers available in the course, which you can use to postpone an obligation to the future.
* Using a Joker, you can move the solving window of a homework assignment forward by $\leq$ 14 days.
* Using a Joker **is not possible if you have already opened the assignment**.
* Before the start of the solving window, one Joker is needed to move the assignment; during the solving window, two are needed.

</details>
